# Les méthodes de cohérence, reproduites depuis leurs formules

Ce carnet rejoue la section 7.3 : ce que valent les méthodes de cohérence
publiées, réimplémentées depuis leurs formules et évaluées sur les mêmes
bâtiments.

Table employée : `coherence_methods_pooled_auc.csv`, l'AUC groupée et centrée par
zone sur quinze zones syriennes — 110 159 bâtiments et 2 247 signalements.

```python
import os
import pandas as pd

# les carnets se lancent depuis notebooks/ ; les tables sont dans data/results
RESULTATS = os.path.join("..", "data", "results")
TABLES = os.path.join("..", "tables")

def lire(dossier, nom, **kw):
    """Lit une table du depot. Le separateur et la virgule decimale varient
    d'une campagne a l'autre : on laisse pandas les deviner sauf indication."""
    kw.setdefault("sep", None)
    kw.setdefault("engine", "python")
    return pd.read_csv(os.path.join(dossier, nom), **kw)

def presque(a, b, tol=0.0006):
    """Egalite au millieme pres, tolerance par defaut d'un demi-millieme."""
    assert abs(a - b) < tol, "%.4f attendu, %.4f obtenu" % (b, a)
    return True
```

Les onze variantes reproduites, de la meilleure à la moins bonne.

```python
c = lire(TABLES, "coherence_methods_pooled_auc.csv", decimal=",")
c.columns = ["methode", "auc"]
print(c.to_string(index=False))
```

La cohérence est un second signal, pas une méthode à elle seule.

```python
# le constat central : la coherence seule classe a peine mieux que le hasard
meilleure = c.auc.max()
print("meilleure methode reproduite : %.3f" % meilleure)
print("ecart au hasard              : %+.3f" % (meilleure - 0.5))
assert meilleure < 0.62, "la coherence seule ne devrait pas depasser 0,62 ici"
print("\nVerifie : la meilleure reproduction reste sous 0,62.")
```

Deux contrôles : la perte due au binaire, et le témoin pré-événement.

```python
# la sortie binaire publiee, et la coherence pre-evenement comme temoin
binaire = c[c.methode.str.contains("binaire")].auc.iloc[0]
temoin = c[c.methode.str.contains("Cohérence pré|pre-sismique|pré-sismique",
                                  regex=True)].auc
print("sortie binaire publiee   : %.3f" % binaire)
if len(temoin):
    print("coherence pre-evenement  : %.3f  (temoin : ne peut rien savoir"
          " de l'evenement)" % temoin.iloc[0])
assert binaire < 0.5, "une sortie binaire devrait perdre le classement"
print("\nVerifie : la sortie binaire publiee passe sous le hasard, ce qui")
print("mesure le cout du passage d'un score continu a une declaration.")
```
